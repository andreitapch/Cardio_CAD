```python
def plot_metrics_comparison(df, title):
    plt.figure(figsize=(10, 6))
    sns.lineplot(x="Handling Method", y="MCC", data=df, label="MCC", marker="o")
    sns.lineplot(x="Handling Method", y="Precision", data=df, label="Precision", marker="o")
    sns.lineplot(x="Handling Method", y="Recall", data=df, label="Recall", marker="o")
    sns.lineplot(x="Handling Method", y="F1", data=df, label="F1", marker="o")
    sns.lineplot(x="Handling Method", y="Accuracy", data=df, label="Accuracy", marker="o")
    sns.lineplot(x="Handling Method", y="ROC-AUC", data=df, label="ROC-AUC", marker="o")
    plt.title(title)
    plt.xlabel('Handling Method')
    plt.ylabel('Metric Value')
    plt.legend(loc="best")
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()
```

```python
# Plot for Imputed Yes
plot_metrics_comparison(df_imputed_yes, "KNN Classifier Metrics (Imputed: Yes)")
```

```python
# Plot for Imputed No
plot_metrics_comparison(df_imputed_no, "KNN Classifier Metrics (Imputed: No)")
```

```python
import plotly.express as px

# Function to plot comparison for Imputed Yes and No using Plotly
def plot_metrics_comparison_plotly(df, title):
    # Melt the DataFrame to make it compatible with Plotly express
    df_melted = df.melt(id_vars=["Handling Method"], value_vars=["MCC", "Precision", "Recall", "F1", "Accuracy", "ROC-AUC"], 
                        var_name="Metric", value_name="Value")

    # Create the interactive line plot
    fig = px.line(df_melted, x="Handling Method", y="Value", color="Metric", 
                  markers=True, title=title, 
                  labels={"Handling Method": "Handling Method", "Value": "Metric Value"})

    # Customize the layout
    fig.update_layout(xaxis_title="Handling Method", 
                      yaxis_title="Metric Value", 
                      legend_title="Metric", 
                      title={'text': title, 'x': 0.5},
                      xaxis_tickangle=-45)

    # Show the plot
    fig.show()

# Plot for Imputed Yes
plot_metrics_comparison_plotly(df_imputed_yes, "KNN Classifier Metrics (Imputed: Yes)")

# Plot for Imputed No
plot_metrics_comparison_plotly(df_imputed_no, "KNN Classifier Metrics (Imputed: No)")
```
